### Lakeflow Connect
- Efficient connectors to ingest data from a wide variety of sources into a Datarbicks Platform

CONNECTORS TYPES 
- Managed COnnectors : out of box connectors
- Standard Connectors : Customizeable Connectors
- Partner Connect : 3rd Party ingestion Services
- Manual File Upload

MANAGED CONNECTORS 

- Low Code, or no Code Ingestion Solutions Fully-Managed by Datarbicks to ingest data from Enterprise applications, Including SAAS Platforms and databases

- out of the Box Connectors that leverage efficient incremental read/write patterns to provide scalable , cost effective , and high Performance data ingestion to the lakehouse

- SAAS COnnectors 
(Saleforce, Servicenow, Workday, Google analytics, Sharepoint)

- DATABASE Connectors Ingest and Sync data from relational databases using change Data Capture
(MYSQL, PostgreSQL, SQL Server)

STANDARD CONNECTORS 
- Customizeable connectors for cloud storage and message buses
- Require writing Custom Python or SQL Code
(Spark Structure Streaming, AUTO Loader or COPY INTO, CREATE TABLE AS SELECT (CTAS))

PARTNER CONNECT 
- Third-party Sevices that support batch or streaming Ingesiton into Datarbicks (Fivertran)

Manual File Upload 
- Manually upload local local files to databricks into either
- Delta Table
- Unity Catalog Volume


MANAGED CONNNECTORS (SAAS INGESTION)

- Serverless , declarative pipelines that incrementally data ingest from publicly accessible sources (Via API,OLAP) to streaming tables
- Managed Ingesiton (Serverless Declarative pipelines)
- SAAS Connectors leverage efficient increemntal reads and writes to make data ingestion faster , scalable and more cost efficeint 
- all data movements in data plane
- control plane is only used for pipeline setup, management and monitoring

DATABASE INGESTION 
- Ingestion Gateway (Classic Compute Declarative pipelines) , Staging and State mangement (Unit caltlog volume)
- Database connectors sync row level changes (insert, update, delete) using Change data Capture (CDC)
- keep up to date , without full database reload


STANDARD Connectors in Lakeflow Connect

- Require writing custom Python and SQL Code
- Spark Structure Streaming : to incrementally ingest data from Apache kafka, Amazon Kinesis
- AutoLoader or Copy INTO : to incremnetally ingest files from Cloud storage (Such as Amazon S3, ADLS, or Unity Catalog Volumes)
- CREATE TABLE AS (CTAS) : To Ingest data from Schema objects such as tables and Views

INGESTION Methods 

- Batch (Load data as batches of rows, often based on schedule), reprocess all records with each run
- Incremntal Batch (Only new data ingested, Faster and efficient ingestion), Spark Structure Streaming & AutoLoader
- Streaming (Continously load micro batches of data at very short, frequent intervals) ideal for Streaming sources such as kafka, Amazon kinesis , Google Pub/sub. 
Spark structure streaming in continouous mode (spark.readStream & AutLoader)
Streaming tables in contnousous pipeline mode
